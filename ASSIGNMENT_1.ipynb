```python
#write a program to check whether a character is an uppercase letter, lowercase letter, digit or special character

ch = input("Enter a single character: ")

if len(ch) != 1:
    print("Please enter only one character.")
elif 'A' <= ch <= 'Z':
    print("The character is an uppercase letter.")
elif 'a' <= ch <= 'z':
    print("The character is a lowercase letter.")
elif '0' <= ch <= '9':
    print("The character is a digit.")
else:
    print("The character is a special character.")
```

```python
#write a program to determine whether a triangle is equilateral, isosceles and scalene

a = float(input("Enter the first side: "))
b = float(input("Enter the second side: "))
c = float(input("Enter the third side: "))

# Check if the sides can form a triangle
if (a + b > c) and (a + c > b) and (b + c > a):
    if a == b == c:
        print("The triangle is Equilateral.")
    elif a == b or b == c or a == c:
        print("The triangle is Isosceles.")
    else:
        print("The triangle is Scalene.")
else:
    print("The given sides do not form a valid triangle.")
```

```python
#write a program to create a simple calculator using if-elif-else

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print("\nChoose an operation:")
print("1. Addition (+)")
print("2. Subtraction (-)")
print("3. Multiplication (*)")
print("4. Division (/)")

choice = input("Enter your choice (+, -, *, /): ")

if choice == "+":
    print("Result =", num1 + num2)
elif choice == "-":
    print("Result =", num1 - num2)
elif choice == "*":
    print("Result =", num1 * num2)
elif choice == "/":
    if num2 != 0:
        print("Result =", num1 / num2)
    else:
        print("Error! Division by zero is not allowed.")
else:
    print("Invalid choice!")
```

```python
#write a program to determine whether a point lies in first quadrant, second quadrant, third quadrant, fourth quadrant, on axis or at origin

x = float(input("Enter x-coordinate: "))
y = float(input("Enter y-coordinate: "))

if x > 0 and y > 0:
    print("The point lies in the First Quadrant.")
elif x < 0 and y > 0:
    print("The point lies in the Second Quadrant.")
elif x < 0 and y < 0:
    print("The point lies in the Third Quadrant.")
elif x > 0 and y < 0:
    print("The point lies in the Fourth Quadrant.")
elif x == 0 and y == 0:
    print("The point is at the Origin.")
elif x == 0:
    print("The point lies on the Y-axis.")
elif y == 0:
    print("The point lies on the X-axis.")
```

##QUESTIONS ON DICTIONARY AND SET


```python
#write a program to create a dictionary from two lists : one of keys and one of values

keys = input("Enter the keys separated by spaces: ").split()
values = input("Enter the values separated by spaces: ").split()

# Check if both lists have the same length
if len(keys) == len(values):
    my_dict = dict(zip(keys, values))
    print("Dictionary:", my_dict)
else:
    print("Error! The number of keys and values must be the same.")
```

```python
#write a program to merge two dictionaries into one

dict1 = {}

n1 = int(input("Enter the number of items in the first dictionary: "))
print("Enter key-value pairs for the first dictionary:")
for i in range(n1):
    key = input("Key: ")
    value = input("Value: ")
    dict1[key] = value

dict2 = {}

n2 = int(input("\nEnter the number of items in the second dictionary: "))
print("Enter key-value pairs for the second dictionary:")
for i in range(n2):
    key = input("Key: ")
    value = input("Value: ")
    dict2[key] = value

# Merge dictionaries
merged_dict = {**dict1, **dict2}

print("\nMerged Dictionary:")
print(merged_dict)
```

```python
#write a program to sort a dictionary by its value

my_dict = {
    "Apple": 50,
    "Banana": 20,
    "Mango": 40,
    "Orange": 30
}

# Sort in ascending order of values
sorted_items = sorted(my_dict.items(), key=lambda x: x[1])

print("Dictionary sorted by values:")
for key, value in sorted_items:
    print(key, ":", value)
```

```python
#write a program to check whether two lists have a least one common element using sets

list1 = list(map(int, input("Enter elements of first list separated by spaces: ").split()))
list2 = list(map(int, input("Enter elements of second list separated by spaces: ").split()))

# Convert lists to sets and find common elements
common = set(list1) & set(list2)

if common:
    print("The two lists have common element(s):", common)
else:
    print("The two lists do not have any common elements.")
```

```python
#write a program to remove duplicate elements from a lsit

numbers = list(map(int, input("Enter the list elements separated by spaces: ").split()))

unique_list = list(set(numbers))

print("List after removing duplicates:")
print(unique_list)
```

```python
#write a program to reverse a list without using built - in reverse functions

numbers = list(map(int, input("Enter the list elements separated by spaces: ").split()))

reversed_list = []

# Traverse the list from the last element to the first
for i in range(len(numbers) - 1, -1, -1):
    reversed_list.append(numbers[i])

print("Reversed List:")
print(reversed_list)
```

```python
#write a program to count even and odd numbers in a list

numbers = list(map(int, input("Enter the list elements separated by spaces: ").split()))

even_count = 0
odd_count = 0

for num in numbers:
    if num % 2 == 0:
        even_count += 1
    else:
        odd_count += 1

print("Number of even elements:", even_count)
print("Number of odd elements:", odd_count)
```

```python
#write a program to merge two lists and sort the final list

list1 = list(map(int, input("Enter elements of first list separated by spaces: ").split()))
list2 = list(map(int, input("Enter elements of second list separated by spaces: ").split()))

# Merge the lists
merged_list = list1 + list2

# Sort the merged list
merged_list.sort()

print("Merged and Sorted List:")
print(merged_list)
```

```python
#write a program to find the second largest element in a list

numbers = list(map(int, input("Enter the list elements separated by spaces: ").split()))

# Remove duplicate elements and sort the list
unique_numbers = list(set(numbers))
unique_numbers.sort()

if len(unique_numbers) < 2:
    print("There is no second largest element.")
else:
    print("Second largest element:", unique_numbers[-2])
```

```python
#write a program to sort a list of tuples based on tuple values

tuples_list = []

n = int(input("Enter the number of tuples: "))

for i in range(n):
    first = input("Enter first value: ")
    second = int(input("Enter second value: "))
    tuples_list.append((first, second))

# Sort the list based on the second value of each tuple
sorted_list = sorted(tuples_list, key=lambda x: x[1])

print("Sorted list of tuples:")
print(sorted_list)
```
