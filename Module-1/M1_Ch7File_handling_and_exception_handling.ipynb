##Read(r)

1. Write a Python program to open a text file named kanifile.txt in read mode, read its complete contents, display the contents and their data type, and then close the file.

```python
f = open("kanifile.txt","r")
data = f.read()
print(data)
print(type(data))
f.close()
```

```python
f = open("randomfile.txt","r")
data = f.read()
print(data)
print(type(data))
f.close()
```

3. Write a Python program to open kanifile.txt in read mode and read the first 18 characters from the file using the read() method. Display the data and its type.

```python
f = open("kanifile.txt","r")
data = f.read(18)
print(data)
print(type(data))
f.close()
```

4. Write a Python program to open kanifile.txt in read mode and read the first three lines of the file using the readline() method.

```python
f = open("kanifile.txt","r")
line1 = f.readline()
print(line1)
line2 = f.readline()
print(line2)
line3 = f.readline()
print(line3)
```

5. Write a Python program to open a text file kanifile.txt in read mode, read the first three lines, remove extra spaces/newline characters using the strip() method

```python
#strip method
f = open("kanifile.txt","r")
line1 = f.readline().strip()
print(line1)
line2 = f.readline().strip()
print(line2)
line3 = f.readline().strip()
print(line3)
```

6. Write a Python program to read the first three lines of a text file and print them without adding extra blank lines between the lines. Use the end="" parameter.

```python
f = open("kanifile.txt","r")
line1 = f.readline()
print(line1, end="")
line2 = f.readline()
print(line2, end="")
line3 = f.readline()
print(line3, end="")
```

8. Write a Python program to demonstrate the use of the with statement in file handling. Open randomfile.txt in read mode using with, read the complete contents of the file using read(), and display the contents.

```python
#with the help of WITH keyword
with open("randomfile.txt", "r") as f:
  data = f.read()
  print(data)
```

#Exception Handling

9. IWrite a Python program to demonstrate exception handling using try, except, and finally. Handle the ZeroDivisionError that occurs when dividing 10 by 0, print "error" when the exception occurs, and always print "execute" using the finally block.

```python
try:
  print(10//0)
except ZeroDivisionError:
  print("error")
finally:
  print("execute")
```
