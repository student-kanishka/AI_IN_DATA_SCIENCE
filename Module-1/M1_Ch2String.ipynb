#**String** **Methods**

7. Write a program to apply all the string methods

```python
#string methods
s1 = "Kanishka kamboj"
print(s1.upper())             #convert all the characters in capital letters
print(s1.lower())             #convert all the characters in small letters
print(s1.title())             #convert first letteer of all string in capital letter
print(s1.capitalize())        #convert first letter of a line in capital letters
print(s1.swapcase())          #convert capital into small and small into capital letters
print(s1.find("h"))           #find the character in the string
print(s1.index("a"))          #find the indexing of a character in string
print(s1.count("h"))          #find the count of a single string in whole string
print(s1.startswith("i"))     #check the string starts with the letter
print(s1.endswith("ka"))      #check the string ends with the letter
```
