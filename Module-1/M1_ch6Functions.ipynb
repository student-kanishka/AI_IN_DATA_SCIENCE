5. Write a Python function calculator(a, b) that takes two numbers and performs addition, subtraction, multiplication, division, and exponentiation, then returns all five results.

```python
def calculator(a, b):
  Sum = a + b
  Sub = a - b
  mul = a * b
  div = a / b
  coff = a ** b
  return Sum, Sub, mul, div, coff
calculator(2,4)
```

6. Write a Python function avg(a, b, c) that accepts three numbers and returns their average.

```python
def avg(a,b,c):
  Sum = a + b + c
  avg = Sum/3
  return avg
avg(2,3,4)
```

8. Write a program to find the factorial of a number using function  

```python
#WAP to find the factorial of a number
n = int(input("enter the number: "))
def factorial(n):
    fact = 1
    for i in range(1, n+1):
      fact = fact * i
    return fact
factorial(n)
```

9. Write a program to find the factorial of a number using local and global variable

```python
#local and global varial
n=int(input("enter number")) #global variable
def factorial(n):
  fact=1                    #local variable
  for i in range(1,n+1):
    fact=fact*i
  return fact


factorial(n)
```
