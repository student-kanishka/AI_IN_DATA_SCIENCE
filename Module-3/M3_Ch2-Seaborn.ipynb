```python
#Important libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Read the CSV file
df = pd.read_csv("penguins.csv").head(50)      #head(50) is used for size
df.head()
```

```python
#Check the number of rows and column
df.shape
```

```python
#create a lineplot chart
sns.lineplot(x="bill_length_mm", y="bill_depth_mm", data=df)
```

```python
#Hue is used to diffrentiate data based on values
sns.lineplot(x="bill_length_mm", y="bill_depth_mm", data=df, hue = "sex")

#Display the plot
plt.show()
```

```python
#Size is used to change the width of lines on chart
sns.lineplot(x="bill_length_mm", y="bill_depth_mm", data=df, hue = "sex", size = 1)

#Display the plot
plt.show()
```

```python
#palettte is used to chane the colour of lines on chart
sns.lineplot(x="bill_length_mm", y="bill_depth_mm", data=df, hue = "sex", size = 1, palette = 'winter')

#Display the plot
plt.show()
```

```python
#Used markers with the style keyword
sns.lineplot(x="bill_length_mm", y="bill_depth_mm", data=df, hue = "sex", size = 1, palette = 'winter', style = 'sex', markers = ["o", "s"])

#Display the plot
plt.show()
```

```python
#Use dashes keyword to break the lines
sns.lineplot(x="bill_length_mm", y="bill_depth_mm", data=df, hue = "sex", size = 1, palette = 'winter', style = 'sex', markers = ["o", "s"], dashes = False)

#Display the plot
plt.show()
```

```python
#Use legend keyword
sns.lineplot(x="bill_length_mm", y="bill_depth_mm", data=df, hue = "sex", size = 1, palette = 'winter', style = 'sex', markers = ["o", "s"], dashes = False, legend = False)

#Display the plot
plt.show()
```

```python
#Use legend keyword
sns.lineplot(x="bill_length_mm", y="bill_depth_mm", data=df, hue = "sex", size = 1, palette = 'winter', style = 'sex', markers = ["o", "s"], dashes = False, legend = False)

#Set the plot style with visible ticks
sns.set_theme(style="ticks")

#Display the plot
plt.show()
```

```python
#Use legend keyword
sns.lineplot(x="bill_length_mm", y="bill_depth_mm", data=df, hue = "sex", size = 1, palette = 'winter', style = 'sex', markers = ["o", "s"], dashes = False, legend = False)

#Set the plot style with visible darkgrids
sns.set_theme(style="darkgrid")

#Display the plot
plt.show()
```

```python
#Use legend keyword
sns.lineplot(x="bill_length_mm", y="bill_depth_mm", data=df, hue = "sex", size = 1, palette = 'winter', style = 'sex', markers = ["o", "s"], dashes = False, legend = False)

#Set the plot style with visible white
sns.set_theme(style="white")

#Display the plot
plt.show()
```

```python
#Use legend keyword
sns.lineplot(x="bill_length_mm", y="bill_depth_mm", data=df, hue = "sex", size = 1, palette = 'winter', style = 'sex', markers = ["o", "s"], dashes = False, legend = False)

#Set the plot style with visible dark
sns.set_theme(style="dark")

#Display the plot
plt.show()
```

```python
#Use legend keyword
sns.lineplot(x="bill_length_mm", y="bill_depth_mm", data=df, hue = "sex", size = 1, palette = 'winter', style = 'sex', markers = ["o", "s"], dashes = False, legend = False)

#Set the plot style with visible whitegrid
sns.set_theme(style="whitegrid")

#Display the plot
plt.show()
```

```python
#Important libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Read the CSV file
df = pd.read_csv("penguins (1).csv").head(50)      #head(50) is used for size
df.head()
```

```python
#Bell Shaped dta aur curve
sns.displot(df["flipper_length_mm"], bins =[175,180,185,190,195],kde = True, rug = True, color =  'pink')

#Display the plot
plt.show()
```

**3.2.3 KDE (Kernel Density Estimations)**

```python
sns.kdeplot(df["flipper_length_mm"], shade = True, color = 'pink', alpha = 0.6)

#Add a title
plt.title('KDE Plot')

#Display the plot
plt.show()
```

**3.3 Box Plot**

```python
sns.boxplot( x = 'flipper_length_mm', y = 'species', data = df, color = 'pink')

#Display the plot
plt.show()
```

**3.4 Violin Plot**

```python
#Used to draw the violin plot
sns.violinplot(y = df['bill_length_mm'])

#Display the plot
plt.show()
```

```python
#Used to draw a violin plot
sns.violinplot(x = df['body_mass_g'], y = df['island'], color = 'g')

#Add a title to violin plot
plt.title('Violin Plot')

#Display the violin plot
plt.show()
```

**3.5 Pair Plot**

```python
#Used to draw a pair plot
sns.pairplot(df, height = 1.5, hue = 'sex', palette = 'winter', diag_kind = 'hist')

#Display the pair plot
plt.show()
```

**3.6 Heatmap**

```python
#Used to draw a Heatmap
matrix = df[['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']].corr()

sns.heatmap(matrix)

#Display the heatmap
plt.show()
```

```python
#Used to draw a Heatmap
matrix = df[['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']].corr()

sns.heatmap(matrix, annot = True, cmap = 'coolwarm')

plt.show()
```
