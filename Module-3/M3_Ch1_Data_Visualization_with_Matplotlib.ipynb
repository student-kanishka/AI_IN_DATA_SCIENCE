3.2) **Histogram**

```python
#Store age
age = [10,20,30,20,10,20,30,40,50,50]

#Creating an historam
plt.hist(age)

#Add a title to the histogram
plt.title("Histogram")

#Add label to x-axis
plt.xlabel("Age")

#Add label to y-axis
plt.ylabel("Frequency")

#Display the histogram
plt.show()
```

```python
#Store age
age = [10,20,30,20,10,20,30,40,50,50,50,60]

#Create a histogram on x-axis and y-axis
plt.hist(age,bins =3,edgecolor = "r",color="y",alpha = 0.6)

#Display grid on histogram
plt.grid(axis='y')

#Add title on the histogram
plt.title("Histogram")

#Add label to x-axis
plt.xlabel("Age")

#Add label to y-axis
plt.ylabel("Frequency")

#Display the line chart
plt.show()
```

```python
#Import numpy library
import numpy as np

#Store age in a list
age = [20, 19, 8, 19, 16, 20, 21, 23, 28, 19, 23]

#Create a histogram and customize it with color, alpha, edgecolor, bins
plt.hist(age, bins=3, edgecolor='r', color='yellow',alpha=0.5)

# Add a title to the histogram
plt.title("Age comparison", fontsize=22, color="green")

# Add a label to the x-axis
plt.xlabel("age", fontsize=12, color="red")

# Add horizontal grid lines
plt.grid(axis='y')

# Add a label to the y-axis
plt.ylabel("", fontsize=12, color="red")

#Calculate the median of the age values
median_age = np.median(age)

#Add a vertical line to show the median age
plt.axvline(median_age,color='blue',linestyle='--',linewidth=3.5,label=f'Median = {median_age}')

#Display the legend
plt.legend()

#Display th histogram
plt.show()
```

**3.3 Bar Chart**

```python
#Store subjects and store
subjects=["python","c","java","DSA"]
score=[40,60,65,85]

#Create a bar chart and customie it with color
plt.bar(subjects,score,color=["r","green","y","pink"])

#Add a title and customize the title
plt.title("Subject vs Score",color='r')

#Add a label to x-axis and customize it with color
plt.xlabel("Subjects",color='r')

#Add a label to y-axis and customize it with color
plt.ylabel("Score",color='r')

#Display the grid
plt.grid(axis='y')

#Display the output
plt.show()
```

3.4) **Pie Chart**

```python
#Store subject and sccore
subject=["Techical","english","Gk","apptitude"]
score=[40,60,65,85]

#Create a pie chart and customize it
plt.pie(score,labels=subject,autopct="%1.2f%%")

#Add a title on pie chart and customize it with color
plt.title("Subject vs Score",color='r')

#Display the pie chart
plt.show()
```

```python
#Store subject and score
subject=["Techical","english","Gk","apptitude"]
score=[40,60,65,85]

#Create a pie chart and customize it
plt.pie(score,labels=subject,autopct="%1.1f%%",startangle=55)

#Add title on pie chart and customize it with color
plt.title("Subject vs Score",color='r')

#Display the pie chart
plt.show()
```

**3.5 Scatter Plot**

```python
import matplotlib.pyplot as plt

#Store variales
x1 = [2, 4, 6, 8, 100]
y1 = [11, 13, 15, 17, 19]

#store two more ariables
x2 = [12, 14, 16, 18, 90]
y2 = [1, 3, 5, 7, 9]

#Add colour and size
rang = ["pink", "purple", "orange", 'red', "green"]
size = [100, 200, 300, 400, 500]

#Create scatter plot
plt.scatter(x1, y1, color = rang, s = size, alpha = 0.5)
plt.scatter(x2, y2, color = rang, s = size, alpha = 0.8)

#Display the legend
plt.legend()

#Display the scatter plot
plt.show()
```
