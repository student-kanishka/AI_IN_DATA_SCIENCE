#**Introduction to statistics in Data Science**


**Descriptive statistics**

```python
#Import libraries
import statistics
import numpy as np
import pandas as pd
```

```python
#Create a list of marks
marks = [10,14,16,20,28,32]
```

```python
#Calculate Mean
mean = statistics.mean(marks)
print("Mean:",mean)

#Calculate Median
median = statistics.median(marks)
print("Median:",median)

#Calculate Mode
mode = statistics.mode(marks)
print("Mode:",mode)
```

```python
#Calculate Variance
variance = statistics.variance(marks)
print("Variance:",variance)

#Calculate Standard Deviation
stdev = statistics.stdev(marks)
print("Stdev:",stdev)
```

```python
#Calculate Marks range
marks_range=max(marks)-min(marks)
print("Marks Range:",marks_range)
```

**IQR - Inter quartile range**

```python
#Calculate IQR
data = np.array( [10,14,16,20,28,32,3000])
Q1 = np.percentile(data,25)
Q3 = np.percentile(data,75)
IQR= Q3-Q1
print("IQR:",IQR)
```

```python
#Calculate lower bound
lower = Q1-1.5*(IQR)
print("Lower:",lower)

#Calculate upper bound
upper = Q3+1.5*(IQR)
print("Upper:",upper)
```

```python
#Identify Outliers
outliers = data[(data<lower)| (data>upper)]
print("Outliers:",outliers)
```

```python
#Display All Statistical results
print("Mean:",mean)
print("Median:",median)
print("Mode:",mode)
print("Variance:",variance)
print("Stdev:",stdev)
print("Marks Range:",marks_range)
print("IQR:",IQR)
print("Lower:",lower)
print("Upper:",upper)
print("Outliers:",outliers)
```
